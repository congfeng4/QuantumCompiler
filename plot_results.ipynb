```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./result/20Q_gate_Tokyo.csv')
df
```

```python
df[df.circuit_name == '20Q_gate_Tokyo_large_1_0_1.5_no.3']
```

```python
df.describe()
```

```python
df.circuit_name.unique()
```

```python
circ_name = 'grover_5'
df_sub = df[df.circuit_name == circ_name]

plt.figure(figsize=(12, 5))
plt.subplot(121)
sns.barplot(data=df_sub, x='graph_model', y='gate_ratio', hue='opt_method')
plt.title(circ_name)

plt.subplot(122)
sns.barplot(data=df_sub, x='graph_model', y='depth_ratio', hue='opt_method')
plt.title(circ_name)
```
