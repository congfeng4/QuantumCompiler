```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(3)
qc.cx(1, 2)      # step 0：依赖 q1, q2
qc.cx(0, 1)      # step 1：依赖 q0, q1
```

```python
for i, inst in enumerate(qc.data):
    print(i, inst.operation.num_qubits, [q._index for q in inst.qubits])
```

```python
from qiskit.converters import circuit_to_dag

dag = circuit_to_dag(qc)
for node in dag.topological_op_nodes():
    print(node.op.name, [q._index for q in node.qargs])
```
