```python
# test_ops_embedding.py
import torch, numpy as np
from contrib.feature_extractor import OpsEmbedding
from contrib.state_space import GateType, OpRepPosition, RoutedStatus

qubit_num = 3
params = {"gate_type_embed_dim": 4, "qubit_embed_dim": 4}
embed = OpsEmbedding(qubit_num, params)

# 伪造 2 条序列（长度不同）
seq1 = torch.tensor([
    [GateType.H,  RoutedStatus.UNROUNTED, 1, 0, 0, 0],
    [GateType.CX, RoutedStatus.UNROUNTED, 1, 1, 0, 1],
    [GateType.SWAP, RoutedStatus.UNROUNTED, 1, 1, 1, 2],
], dtype=torch.long)

seq2 = torch.tensor([
    [GateType.CX, RoutedStatus.ROUTED, -1, 2, 0, 1],
], dtype=torch.long)

out1 = embed(seq1)  # [len1+CLS, hidden]
out2 = embed(seq2)  # [len2+CLS, hidden]

print("seq1 shape:", seq1.shape, "→ embed out:", out1.shape)
print("seq2 shape:", seq2.shape, "→ embed out:", out2.shape)
print("hidden_size =", embed.hidden_size)
```
