```python
from tinydb import TinyDB, Query

db = TinyDB("my.json")
User = Query()
db.insert({"name": "Alice", "age": 23})
db.insert({"name": "Bob", "age": 25})

# Mongo-like 链式查询
res = db.search(User.age > 23)
print(res)   # [{'name': 'Bob', 'age': 25}]
```

```python
db.contains(User.name == 'Alice')
```

```python
import shelve

db = shelve.open("my.shv", writeback=True)   # writeback 自动回写
db["counter"] = 0
db["counter"] += 1
db["user"] = {"name": "Alice"}   # 值必须是 pickle-able
print(db["counter"])             # 2
```
