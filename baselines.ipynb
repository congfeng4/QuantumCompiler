```python
from contrib.random_graphs import generate_graph_for_num_qubits
from qiskit.converters import circuit_to_dag
from qiskit import QuantumCircuit
```

```python
coupling_map, name = generate_graph_for_num_qubits('star', 4)
```

```python
coupling_map
```

```python
dag = (QuantumCircuit.from_qasm_file('./data/20Q_gate_Tokyo/circuits/20Q_gate_Tokyo_large_1_10_1.5_no.0.qasm'))
```

```python
from qiskit.transpiler import PassManager
from qiskit.transpiler.passes import DenseLayout, ApplyLayout

pm = PassManager([DenseLayout(coupling_map),
                  ApplyLayout()])
pm.run(dag)                      # dag 是中间表示
init_layout = pm.property_set['layout']   # 同 1-A
print("DenseLayout result:", init_layout)
```
