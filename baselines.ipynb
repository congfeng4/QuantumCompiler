```python
import pandas as pd
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit.transpiler import CouplingMap, Target
from qiskit.transpiler import PassManager

from contrib.common import get_total_ops

# 示例：生成 level=3 的 PassManager
pm = generate_preset_pass_manager(
    optimization_level=3,
    # target=Target.from_configuration({'gates': ['cx', 'rz', 'sx', 'x']}),
    coupling_map=CouplingMap.from_line(5)
)
```

```python
from qiskit import QuantumCircuit


def qiskit_optimize(qc: QuantumCircuit, level: int):
    pm = generate_preset_pass_manager(optimization_level=level).optimization
    return pm.run(qc)
```

```python
qc_in = QuantumCircuit.from_qasm_file('./data/nam_circs/barenco_tof_3.qasm')
qc_out = qiskit_optimize(qc_in, level=3)
```
